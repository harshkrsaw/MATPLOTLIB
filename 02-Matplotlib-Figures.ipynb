```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
x = np.arange(10)
y = x *2

a = x
b = x**4
```

```python
fig = plt.figure()

axes1 = fig.add_axes([0,0,1,1])

axes1.plot(x,y)
axes1.set_xlim(0,9)
axes1.set_ylim(0,18)
axes1.set_title('x - x*2')
axes1.set_xlabel('X - axis')
axes1.set_ylabel('Y - axis')


axes2 = fig.add_axes([0.6,0.1,0.25,0.25])

axes2.plot(x,y)
axes2.set_xlim(1,2)
axes2.set_ylim(2,4)
axes2.set_title('Zoomed In(1,2)')


axes3 = fig.add_axes([0.1,0.6,0.25,0.25])

axes3.plot(a,b)
axes3.set_xlim(0,9)
axes3.set_ylim(0,6700)
axes3.set_title('x - x**4')
```

```python
fig2 = plt.figure()
axes1 = fig2.add_axes([0,0,1,1,])

axes2 = fig2.add_axes([1,1,0.5,0.5])
```

### Part 3 - Figure Parameter

```python
#Fig size, dpi and savefig

fig = plt.figure(figsize=(10,4),dpi=100)

axes1 = fig.add_axes([0,0,1,1])

axes1.plot(a,b)

fig.savefig('new_fig.png',bbox_inches = 'tight')
```
