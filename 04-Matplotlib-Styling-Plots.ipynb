```python
import matplotlib.pyplot as plt
import numpy as np
```

### Adding legend and editing it

```python
a = np.linspace(0,10,11)
```

```python
fig = plt.figure()

axes = fig.add_axes([0,0,1,1])
axes.plot(a,a,label = 'X vs X')

axes.plot(a,a**2,label = 'X vs X^2')

#Adding the legend
axes.legend()
```

```python
fig = plt.figure()

axes = fig.add_axes([0,0,1,1])
axes.plot(a,a,label = 'X vs X')

axes.plot(a,a**2,label = 'X vs X^2')

#Editing the legend
axes.legend(loc = (0.1,0.8))
```

```python
fig = plt.figure()

axes = fig.add_axes([0,0,1,1])
axes.plot(a,a,label = 'X vs X')

axes.plot(a,a**2,label = 'X vs X^2')

#Editing the legend
axes.legend(loc = (1.2,0.5))
```

### Adding Colours

```python
fig = plt.figure()

axes = fig.add_axes([0,0,1,1])
axes.plot(a,a,color = 'green', label = 'X vs X')

axes.plot(a,a+2,color = 'red',label = 'X vs X+2')

axes.legend()

# Or we can add custom color as well while passing RGB HEX code (which we can fing from ggogle).
```

### Line Width

```python
fig = plt.figure()

axes = fig.add_axes([0,0,1,1])
axes.plot(a,a,color = 'green',lw = 10) # Adding linewidth (we can choose any number)

axes.plot(a,a+2,color = 'red',linewidth = 10) # Adding linewidth
```

### Line Style

```python
fig = plt.figure()

axes = fig.add_axes([0,0,1,1])


# possible linestype options '-', '--', '-.', ':', ''
axes.plot(a,a,color = 'green',ls = '--') 
# We can choose any of the options above and for more options we can refer to stackover flow or official link.
# We can create a custome line style as well for that refer to official notebook

axes.plot(a,a+2,color = 'red',linewidth = 10 ,linestyle = ':') 
```

### Marker

```python
fig = plt.figure()

ax = fig.add_axes([0,0,1,1])

# We have multiple marker options like we can try multiple symbles or we can do googel search for more symble.
# Marker = ".","o","v","s","+","D"
ax.plot(a,a+10,color = 'green',ls = '--',marker = 'o')

# We can customize our marker as well with some more options like markersize, markeredgecolor, markeredgewidth, markerfacecolor, etc. 
#(For more optoins we can check the documentary)
ax.plot(a,a,color = 'red',marker = 'o',markersize = 16,markeredgewidth = 5, markeredgecolor = 'blue',)
```
