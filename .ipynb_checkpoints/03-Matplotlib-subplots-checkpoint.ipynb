```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

# Subplots

```python
a = np.linspace(0,10,11)
b = a*2

x = np.arange(10)
y = x**4
```

```python
fig,axes = plt.subplots(nrows = 2,ncols = 2)

axes[0][0].plot(x,y)
axes[0][1].plot(a,b)
axes[1][0].plot(a,b)
axes[1][1].plot(x,y)

plt.tight_layout() # Fix the spaceing issue
```

```python
fig,axes = plt.subplots(nrows = 3,ncols = 1)
axes[0].plot(x,y)
axes[1].plot(a,b)
axes[2].plot(a,b)

plt.tight_layout()
```

### Manually fixing the layout

```python
fig,axes = plt.subplots(nrows = 2,ncols = 2)

axes[0][0].plot(x,y)
axes[0][1].plot(a,b)
axes[1][0].plot(a,b)
axes[1][1].plot(x,y)

#Manually fixing the layout
fig.subplots_adjust(wspace=0.3,hspace=0.25)
```

### Editing the parameters of subplots

```python
fig,axes = plt.subplots(nrows = 2,ncols = 2)

axes[0][0].plot(x,y)
axes[0][0].set_title('[0] [0]')
axes[0][0].set_ylabel('Adding Y label')


axes[0][1].plot(a,b)
axes[0][1].set_title('[0] [1]')
axes[0][1].set_xlabel('Adding x label')

axes[1][0].plot(a,b)
axes[1][0].set_title('[1] [0]')
axes[1][0].set_ylabel('Adding Y label')

axes[1][1].plot(x,y)
axes[1][1].set_title('[1] [1]')
axes[1][1].set_xlabel('Adding X label')

fig.tight_layout()
```

### Adding a super title

```python
fig,axes = plt.subplots(nrows = 3,ncols = 1)
axes[0].plot(x,y)
axes[1].plot(a,b)
axes[2].plot(a,b)


fig.suptitle('Super title',fontsize = 20)

plt.tight_layout()
```

### Saving the figure

```python
fig,axes = plt.subplots(nrows = 2,ncols = 2)

axes[0][0].plot(x,y)
axes[0][0].set_title('[0] [0]')
axes[0][0].set_ylabel('Adding Y label')


axes[0][1].plot(a,b)
axes[0][1].set_title('[0] [1]')
axes[0][1].set_xlabel('Adding x label')

axes[1][0].plot(a,b)
axes[1][0].set_title('[1] [0]')
axes[1][0].set_ylabel('Adding Y label')
axes[1][0].set_xlabel("Adding X label")

axes[1][1].plot(x,y)
axes[1][1].set_title('[1] [1]')
axes[1][1].set_xlabel('Adding X label')
axes[1][1].set_ylabel('Adding Y label')

fig.tight_layout()




# Saving the figure
fig.savefig('newfig.png',bbox_inches = 'tight')
```
